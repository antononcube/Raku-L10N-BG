## Translate "RU.l10n"

Read the Russian L10N localization file:

```raku
my $fileName = $*CWD.add(<.. RU.l10n>);
say "file exists : {$fileName.f}";
my $ruCode = data-import($fileName, 'asis');
text-stats($ruCode)
```

Translate the Russian localization to a Bulgarian localization using an LLM:

```raku
#`[
my $bgCode = llm-synthesize([
    "Translate the following L10N Raku keywords file from Russian to Bulgarian:\n\n",
    $ruCode,
    "\n\n",
    llm-prompt('NothingElse')('L10N-spec')
    ],
    e => $conf-terra
);

text-stats($bgCode)
]
```

Export the obtained "BG.l10n" code (to a file with suffix "-new"):

```raku
spurt($*CWD.add(<.. BG.l10n-new>), $bgCode)
```
